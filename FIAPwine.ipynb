```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(16, 8))
sns.heatmap(wdRed2.corr(), annot=True, cmap="inferno")
```

```python
plt.figure(figsize=(16, 8))
plt.title("Distribuição de acidez volátil por qualidade de vinho")
plt.xlabel("Qualidade do vinho")
plt.ylabel("Acidez volátil")
sns.lineplot(data=wdRed,x = 'quality', y ='volatile acidity')
plt.show()
```

```python
plt.figure(figsize=(16, 8))
plt.title("Distribuição do teor alcóolico por qualidade de vinho")
plt.xlabel("Qualidade do vinho")
plt.ylabel("Teor álcoolico")
sns.lineplot(data=wdRed,x = 'quality', y ='alcohol')
plt.show()
```

```python

#Relação entre acidez volátil e teor álcoolico e relação entre teor álcoolico e qualidade do vinho.
corr1 = wdRed["alcohol"].corr(wdRed["volatile acidity"])
corr2 = wdRed["alcohol"].corr(wdRed["quality"])

data = {
    'variavel1' : ["alcohol", "alcohol"],
    'variavel2' : ["volatile acidity" ,"quality"],
    'correlação' : [round(corr1, 2), round(corr2, 2) ]
}

wncorr = pd.DataFrame(data)
wncorr
```

```python
wdRedHighq = wdRed[wdRed["quality"] >= 7]
wdRedHighq
```

```python
# Qual a acidez volátil média dos vinhos de qualidade elevada?
meanalc = wdRedHighq["volatile acidity"].mean()
print(f"Média de acidez volátil dos vinhos de boa qualidade é de {round(meanalc, 1)}g/L")
```

```python
# https://www.blog.psicometriaonline.com.br/o-que-e-correlacao-de-pearson/
# Qual a relação entre teor álcoolico e qualidade do vinho?

corralc = wdRed["alcohol"].corr(wdRed["quality"])
print(f"A correlação entre teor álcoolico e qualidade do vinho é de {round(corralc, 2)}, sendo considerada moderada para Cohen e fraca para Rumsey")
```

```python
# Há alguma relação entre acidez volátil e teor álcoolico?
corrvol = wdRed["alcohol"].corr(wdRed["volatile acidity"])
print(f"A correlação entre acidez volátil e teor álcoolico é de {round(corrvol, 2)}, dando a entender que são inversamente proporcionais e fracas")
```

```python
# Há alguma relação entre acidez volátil e qualidade?
corrvolq = wdRed["volatile acidity"].corr(wdRed["quality"])
print(f"A correlação entre acidez volátil e qualidade é de {round(corrvolq, 2)}, dando a entender que são inversamente proporcionais e fracas")
```
